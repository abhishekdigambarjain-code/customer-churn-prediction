```python
X = df.drop(columns=["Churn", "customerID"])
y = df["Churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
numerical_cols = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Numerical columns:")
print(numerical_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

```python
numerical_cols = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

categorical_cols = [
    col for col in X_train.columns
    if col not in numerical_cols
]

print("Numerical columns:")
print(numerical_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_cols
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_cols
        )
    ]
)

print("Preprocessor created successfully")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numerical_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
])

print("Preprocessor created successfully")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])

print("Logistic Regression pipeline created successfully")
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully")
```

```python
y_pred = model.predict(X_test)

print(y_pred[:20])
```

```python
from sklearn.metrics import confusion_matrix, classification_report

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
balanced_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

balanced_model.fit(X_train, y_train)

print("Balanced model training completed successfully")
```

```python
balanced_pred = balanced_model.predict(X_test)

print(balanced_pred[:20])
```

```python
from sklearn.metrics import confusion_matrix, classification_report

balanced_cm = confusion_matrix(y_test, balanced_pred)

print("Balanced Model - Confusion Matrix:")
print(balanced_cm)

print("\nBalanced Model - Classification Report:")
print(classification_report(y_test, balanced_pred))
```

```python
from sklearn.metrics import roc_auc_score

# Probability of Churn = 1
original_prob = model.predict_proba(X_test)[:, 1]
balanced_prob = balanced_model.predict_proba(X_test)[:, 1]

# ROC-AUC scores
original_auc = roc_auc_score(y_test, original_prob)
balanced_auc = roc_auc_score(y_test, balanced_prob)

print("Original Logistic Regression ROC-AUC:", round(original_auc, 3))
print("Balanced Logistic Regression ROC-AUC:", round(balanced_auc, 3))
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

# ROC values
fpr_original, tpr_original, _ = roc_curve(y_test, original_prob)
fpr_balanced, tpr_balanced, _ = roc_curve(y_test, balanced_prob)

# Plot
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_original,
    tpr_original,
    label=f"Original LR (AUC = {original_auc:.3f})"
)

plt.plot(
    fpr_balanced,
    tpr_balanced,
    label=f"Balanced LR (AUC = {balanced_auc:.3f})"
)

plt.plot([0, 1], [0, 1], "--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Logistic Regression")
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ))
])

rf_model.fit(X_train, y_train)

print("Random Forest training completed successfully")
```

```python
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score

rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest - Confusion Matrix:")
print(confusion_matrix(y_test, rf_pred))

print("\nRandom Forest - Classification Report:")
print(classification_report(y_test, rf_pred))

print("Random Forest ROC-AUC:", round(roc_auc_score(y_test, rf_prob), 3))
```

```python
from xgboost import XGBClassifier

xgb_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        random_state=42,
        eval_metric="logloss"
    ))
])

xgb_model.fit(X_train, y_train)

print("XGBoost training completed successfully")
```

```python
xgb_pred = xgb_model.predict(X_test)
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]

print("XGBoost - Confusion Matrix:")
print(confusion_matrix(y_test, xgb_pred))

print("\nXGBoost - Classification Report:")
print(classification_report(y_test, xgb_pred))

xgb_auc = roc_auc_score(y_test, xgb_prob)

print("XGBoost ROC-AUC:", round(xgb_auc, 3))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50]

for threshold in thresholds:
    
    # Convert probabilities into 0/1 using the threshold
    predictions = (xgb_prob >= threshold).astype(int)

    precision = precision_score(y_test, predictions)
    recall = recall_score(y_test, predictions)
    f1 = f1_score(y_test, predictions)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.2f} | "
        f"Recall: {recall:.2f} | "
        f"F1: {f1:.2f}"
    )
```

```python
X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training set:", X_train_final.shape)
print("Validation set:", X_val.shape)
print("Test set:", X_test.shape)
```

```python
xgb_model.fit(X_train_final, y_train_final)

print("XGBoost retrained on final training set successfully")
```

```python
xgb_val_prob = xgb_model.predict_proba(X_val)[:, 1]

print(xgb_val_prob[:10])
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50]

for threshold in thresholds:

    val_pred = (xgb_val_prob >= threshold).astype(int)

    precision = precision_score(y_val, val_pred)
    recall = recall_score(y_val, val_pred)
    f1 = f1_score(y_val, val_pred)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.2f} | "
        f"Recall: {recall:.2f} | "
        f"F1: {f1:.2f}"
    )
```

```python
# Get probabilities on untouched test data
final_test_prob = xgb_model.predict_proba(X_test)[:, 1]

# Apply threshold selected using validation data
final_test_pred = (final_test_prob >= 0.35).astype(int)

print("Final XGBoost - Confusion Matrix:")
print(confusion_matrix(y_test, final_test_pred))

print("\nFinal XGBoost - Classification Report:")
print(classification_report(y_test, final_test_pred))

final_auc = roc_auc_score(y_test, final_test_prob)
print("Final ROC-AUC:", round(final_auc, 3))
```
